```python
from datetime import datetime
class Expense:
    def __init__(self,amount,category,description):
        self.amount=amount
        self.category=category
        self.description=description
        self.date=datetime.now()
    def __str__(self):
        return (f"rs{self.amount}| "
                f"{self.category}| "
                f"{self.description}| "
                f"{self.date.strftime('%d-%m-%Y')}")
```

```python
class User:
    def __init__(self,name):
        self.name=name
        self.expenses=[]
    def add_expense(self,expense):
        self.expenses.append(expense)
    def show_expenses(self):
        for i in self.expenses:
            print(i)
```

```python
class ExpenseManager:
    def __init__(self):
        self.users={}
    def add_user(self,name):
        self.users[name]=User(name)
    def add_expense(self,username,amount,category,description):
        expense=Expense(amount,category,description)
        self.users[username].add_expense(expense)
    def total_spending(self,username):
        expenses=self.users[username].expenses
        return sum(expense.amount for expense in expenses)
```

```python
manager=ExpenseManager()
manager.add_user("Shraddha")
manager.add_expense("Shraddha",
                    500,
                    "Food","Pizza")
manager.add_expense("Shraddha",200,"Travel","Auto")
manager.users["Shraddha"].show_expenses()
print("Total Spending:", manager.total_spending("Shraddha"))
```
